# 1. import necessary libraries

```python
import pandas as pd

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score
```

# 2. IMPORT DATASET

```python
wine_data = load_wine()
```

```python
wine_data_df = pd.DataFrame(wine_data.data, columns = wine_data.feature_names)
wine_data_df["Target"] = wine_data.target
wine_data_df
```

# 3.data understanding

```python
wine_data_df.shape
```

```python
wine_data_df.dtypes
```

```python
wine_data_df.isna().sum()
```

# 4. Data preprocessing

```python
X = wine_data_df.drop("Target", axis =1)
y = wine_data_df["Target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.20,random_state = 123,shuffle = True, stratify = y)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

```python
std_scalar = StandardScaler()
scaled_X_train = std_scalar.fit_transform(X_train)
```

```python
scaled_X_test = std_scalar.transform(X_test)
```

# 5. MODEL BUILDING || 6. MODEL TRAINING

```python
KNN_Classifier = KNeighborsClassifier(n_neighbors=3)
knn_model = KNN_Classifier.fit(scaled_X_train, y_train)
y_train_pred = knn_model.predict(scaled_X_train)
```

```python
knn_model = KNN_Classifier.fit(scaled_X_test, y_test)
y_test_pred = knn_model.predict(scaled_X_test)
```

# 7.MODEL TESTING

```python
print("Accuracy Score: ", round(accuracy_score(y_train,y_train_pred),4))
```

```python
KNN_Classifier = KNeighborsClassifier(n_neighbors=5)
knn_model = KNN_Classifier.fit(scaled_X_train, y_train)
y_train_pred = knn_model.predict(scaled_X_train)
```

```python
print("Accuracy Score: ", round(accuracy_score(y_train,y_train_pred),4))
```

```python
KNN_Classifier = KNeighborsClassifier(n_neighbors=7)
knn_model = KNN_Classifier.fit(scaled_X_train, y_train)
y_train_pred = knn_model.predict(scaled_X_train)
```

```python
print("Accuracy Score: ", round(accuracy_score(y_train,y_train_pred),4))
```

# using for loop to find the optimal n_neighbours

```python
from sklearn.model_selection import cross_val_score, KFold
```

```python
k_fold = KFold(n_splits=5, shuffle=True, random_state=123)
cv_scores_mean = []

for i in range(3,52,2):
    knn_classifier = KNeighborsClassifier(n_neighbors = i)
    knn_classifier.fit(scaled_X_train, y_train)
    cv_scores = cross_val_score(estimator =knn_classifier,X = scaled_X_train, y = y_train, cv=k_fold )
    cv_scores_mean.append(cv_scores.mean().round(4))
   
```

```python
max(cv_scores_mean)
```

```python
cv_scores_mean.index(max(cv_scores_mean))
```

```python
KNN_Classifier = KNeighborsClassifier(n_neighbors=11)
knn_model = KNN_Classifier.fit(scaled_X_train, y_train)
y_train_pred = knn_model.predict(scaled_X_train)
print("Accuracy Score: ", round(accuracy_score(y_train,y_train_pred),4))
```
