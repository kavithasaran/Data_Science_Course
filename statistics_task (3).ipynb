## 1.	The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance

### type of test & hypothesis:
Statistics Test:
  One sample t-test ( we are comparing one sample mean to a population mean )
  
  Right-tailed test ( we are going to check whether the performance increased)
  
Hypothesis:

    null hypothesis: H0    
        Average = 65
        
    alternate hypothesis:Ha
    
        Average > 65

```python
import numpy as np
np.random.seed(50)
scores = np.random.normal(loc=68, scale=10, size=50)
scores
print("Mean:", scores.mean())
print("Std:", scores.std(ddof=1))
```

```python
from scipy.stats import ttest_1samp
t_stat,p_2_tailed = ttest_1samp(scores,popmean=65)
t_stat,p_2_tailed
```

```python
if t_stat > 0:
    p_1_tailed = p_2_tailed / 2
else:
    p_1_tailed = 1 - (p_2_tailed/2)
print("t-statistic value:", round(t_stat,3))
print("one-tailed p_value:", round(p_1_tailed,3))
```

```python
if p_1_tailed < 0.05:
    print("Null hypothesis can be rejected")
else:
    print("Null hypothesis cannot be rejected")
```

### conclusion statement:
At 5% level of significance, there is sufficient statistical evidence to conclude that the new course improves performance
    

## 2.	A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

right-tailed test (we want to test whether the average resolution time has increased)

Hypothesis:

Null hypothesis: H0

Average resolution time = 30

Alternate hypothesis: Ha

Average resolution time > 30

```python
from scipy.stats import ttest_1samp
t_stat, p_2_tail_value = ttest_1samp(time, popmean = 30)
t_stat, p_2_tail_value
```

```python
if t_stat > 0:
    p_1tailed = p_2_tail_value / 2
else:
    p_1tailed = 1 - (p_2_tail_value / 2)
print("t-stat value:", round(t_stat,3))
print("p_1tailed value:", round(p_1tailed,3))     
```

```python
if p_1tailed < 0.05:
    print("Null hypothesis can be rejected")
else:
    print("Null hypothesis cannot be rejected")
```

### conclusion statement:
At 5% level of significance, there is sufficient statistical evidence that the average resolution time has increased.

## 3.	A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed.

### type of test and hypothesis:
statistics test:

one sample proportion z- test ( we compare a sample proprtion against a population proportion, since sample size is large z-test is used)

right-tailed ( we need to test whether customer satisfaction is higher)

hypothesis:

null hypothesis: H0

p_value = 0.6

alternate hypothesis: Ha

p_value > 0.6


```python
from statsmodels.stats.proportion import proportions_ztest
z_stat, p_2_tailed = proportions_ztest(count = 134,
    nobs = 200,
    value=0.6)
z_stat, p_2_tailed
```

```python
if z_stat > 0:
    p_1_tailed = p_2_tailed / 2 
else:
    p_1_tailed = 1 - (p_2_tailed / 2)
print("z_stat_value:", round(z_stat,3))
print("p_1_tailedvalue:", round(p_1_tailed,4))
```

```python
if p_1_tailed < 0.05:
    print("Null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
```

### conclusion statement:
at 5% level of significance, there is sufficient statistical evidence that the customer satisfaction is higher than 60%

## 5.	Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

## type of test and hypothesis:
statistics test:

2 sample t-test ( samples from two independent groups are compared for mean value)

2 tailed ( need to find if there is any difference)

hypothesis:

null hypothesis: H0

mean of meThod A = mean of method B

Alternate hypothesis: Ha

mean of meThod A NOT= mean of method B

```python
import numpy as np
np.random.seed(42)
method_A = np.random.normal(loc=75, scale=12, size=30)
method_B = np.random.normal(loc=70, scale=10, size=28)
```

```python
from scipy.stats import ttest_ind
t_stat, p_value = ttest_ind(method_A,method_B,equal_var = False)
t_stat, p_value
```

```python
if p_value < 0.05:
    print("Null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
    
```

## conclusion statement:
At 5% level of significance, there is not sufficient statistical evidence that there is a difference in average scores between 2 methods.

## 9.	A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

### type of test and hypothesis:
statistics test:

one way ANOVA (3 independent groups against one category)

hypothesis:

null hypothesis: H0

mean_programA = mean_programB = mean_programC

Alternate hypothesis: Ha

atleast one mean value is significantly different from others

```python
import numpy as np
np.random.seed(42)
program_A = np.random.normal(loc=80, scale=8, size=25)
program_B = np.random.normal(loc=75, scale=7, size=30)
program_C = np.random.normal(loc=82, scale=6, size=28)
```

```python
from scipy.stats import f_oneway
```

```python
f_stat, p_value = f_oneway(program_A,program_B,program_C)
f_stat, p_value
```

```python
if p_value < 0.05:
    print("null hypothesis is rejected")
else:
    print("null hypothesis is not rejected")
    
```

### conclusion statement:
at 5% level of significance, there is sufficient staistical evidence that there is a difference in the mean productivity across 3 programs

## 10.	A company wants to test whether gender and product preference are independent. The survey results are as follows:
●	Male: 60 prefer Product A, 40 prefer Product B
●	Female: 50 prefer Product A, 50 prefer Product B
At the 5% level of significance, test for independence.


## type of test and hypothesis:
statistics test:

Chi-squares test of independence ( testing the relationship between 2 categorical variables, contigency table can be created)

hypothesis:

null hypothesis: H0

no association between 2 categories \ product preference is independent related to categories

alternate hypothesis: Ha

high association between 2 categories \ gender and product preference are highly dependent

```python
import numpy as np
data = np.array([[60,40],[50,50]])
```

```python
from scipy.stats import chi2_contingency
```

```python
statistic,p_value,dof,expected_freq = chi2_contingency(data)
statistic,p_value,dof,expected_freq
```

```python
if p_value < 0.05:
    print("null hypothesis is rejected")
else:
    print("null hypothesis is not rejected")
```

### conclusion statement:
at 5% level of significance, there is sufficient statistical evidence that gender and product preference are independent

## 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
●	Face 1: 18
●	Face 2: 22
●	Face 3: 20
●	Face 4: 25
●	Face 5: 17
●	Face 6: 18
At the 5% level of significance, test whether the die is fair.


### type of test and hypothesis:
statistics test:

chi-squared goodness of fit test ( we want to compare the observed value with expected value , and check whether those fit)

hypothesis:

null hypothesis: H0

the die is fair \ no significant diff between observed value and expected value

alternate hypothesis: Ha

the die is not fair \ significant difference between observed and expected value

```python
import numpy as np
observed = np.array([18,22,20,25,17,18])
expected = np.array([20,20,20,20,20,20])
```

```python
from scipy.stats import chisquare
chi2_stat, p_value = chisquare(observed,expected)
chi2_stat, p_value
```

```python
if p_value < 0.05:
    print("null hypothesis is rejected")
else:
    print("null hypothesis is not rejected")
```

### conclusion statement:
at 5% level of significance, there is sufficient statistical evidence that the die is fair
