```python
import pandas as pd

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.feature_selection import RFE, SelectFromModel

from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
```

```python
breast_cancer = load_breast_cancer()
breast_cancer_df = pd.DataFrame(breast_cancer.data, columns = breast_cancer.feature_names)
breast_cancer_df["Target"] = breast_cancer.target
```

```python
breast_cancer_df
```

```python
X = breast_cancer_df.drop("Target", axis=1)
y = breast_cancer_df["Target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.20, random_state=23,
                                                        shuffle=True,
                                                        stratify=y)
```

```python
rf_model = RandomForestClassifier(max_depth=3)
```

```python
rf_model.fit(X_train,y_train)
```

```python
y_pred_train = rf_model.predict(X_train)
y_pred_test = rf_model.predict(X_test)
```

```python
print("Accuracy score_train :", round(accuracy_score(y_train,y_pred_train),4))
print("Accuracy score_test :", round(accuracy_score(y_test,y_pred_test),4))

print("Precision score_train :", round(precision_score(y_train,y_pred_train),4))
print("Precision score_test :", round(precision_score(y_test,y_pred_test),4))

print("Recall score_train :", round(recall_score(y_train,y_pred_train),4))
print("Recall score_test :", round(recall_score(y_test,y_pred_test),4))

print("Confusion matrix_train :", confusion_matrix(y_train,y_pred_train))
print("Confusion matrix_test :", confusion_matrix(y_test,y_pred_test))
```

```python
rfe_select = RFE(RandomForestClassifier(),n_features_to_select=12)
```

```python
rfe_select.fit(X_train,y_train)
```
