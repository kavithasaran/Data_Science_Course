```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from pickle import dump, load
```

```python
college_data = pd.read_csv("college_list.csv")
college_data
```

```python
X = college_data.drop("College_Name", axis=1)
```

```python
college_data.shape
```

```python
college_data.describe()
```

```python
college_data.dtypes
```

```python
scaler = StandardScaler()
scaled_X = scaler.fit_transform(X)
scaled_X = pd.DataFrame(scaled_X)
scaled_X
```

```python
scaled_X.mean(), scaled_X.std()
```

```python
agg_model = AgglomerativeClustering(n_clusters=4,linkage='ward')
clusters = agg_model.fit_predict(scaled_X)
clusters
```

```python
X["College_level"] = clusters
```

```python
X
```

```python
X[X["College_level"] == 1]
```

```python
X.groupby(by="College_level").agg(['mean'])
```

```python
X.groupby(by="College_level").agg(["min","max"])
```

```python
X
```

```python
cluster_names = {3:"Best college",
                1:"Mid-level college",
                 0:"emerging colleges",
                2:"emerging colleges"}
X["Category"] = X["College_level"].map(cluster_names)
```

```python
X
```

```python
(X.Category).nunique()
```

# now supervised learning

```python
input_data = X.drop(["College_level","Category"], axis=1)
output_data = X["College_level"]
X_train, X_test, y_train,y_test = train_test_split(input_data,output_data,test_size=0.20,random_state=12,
                                                    shuffle=True,
                                                    stratify=output_data)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

```python
rf_model = RandomForestClassifier(max_depth=5)
rf_model.fit(X_train,y_train)
y_pred_train = rf_model.predict(X_train)
y_pred_test = rf_model.predict(X_test)
print("Train Accuracy score :", accuracy_score(y_train,y_pred_train))
print("Test Accuracy score :", accuracy_score(y_test,y_pred_test))
```
