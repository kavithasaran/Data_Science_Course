# 1. Import neccesary libraries

```python
from sklearn import datasets
import pandas as pd
```

# 2. Import dataset

```python
breast_cancer_data = datasets.load_breast_cancer()
```

```python
print(breast_cancer_data.DESCR)
```

```python
breast_cancer_data_df = pd.DataFrame(data = breast_cancer_data.data, columns= breast_cancer_data.feature_names)
breast_cancer_data_df["Target"] = breast_cancer_data.target
breast_cancer_data_df
```

# 3.Data understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.mean()
```

# 4.Data preparation - No actions needed for this data

# 5.Model building

```python
X = breast_cancer_data_df.drop(labels = "Target", axis = 1)
y = breast_cancer_data_df["Target"]
```

```python
X.shape, y.shape
```

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
```

```python
plt.pie(x = y.value_counts(),labels = ["Benign cancer","Malignant cancer"],autopct = "%1.0f%%")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.20,shuffle=True,random_state = 12,stratify=y)
```

```python
plt.pie(x = y_train.value_counts(),labels = ["Benign cancer","Malignant cancer"],autopct = "%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_test.value_counts(),labels = ["Benign cancer","Malignant cancer"],autopct = "%1.0f%%")
plt.show()
```

```python
print("Training samples:", X_train.shape, y_train.shape)
print("Testing samples:", X_test.shape, y_test.shape)
```

# 5.Model Training

```python
%%time
from sklearn.tree import DecisionTreeClassifier
DT_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight = {0:2,1:1})
DT_model
```

```python
DT_model.fit(X_train,y_train)
```

# 6. Model testing & 7.Model Evaluation

## Training data

```python
y_pred_train = DT_model.predict(X_train)
y_pred_train
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

## Test data

```python
y_pred_test = DT_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

# Model OPMIZATION Techniques

## Improve model performance - accuracy score, precision, recall, etc....

### When the units of input features are different, model optimization can be done by optimizing all features


### Lets try with standard scaler =>>>   mean = 0 and std.deviation=1

```python
from sklearn.preprocessing import StandardScaler
std_scalar = StandardScaler()
scaled_X = pd.DataFrame(std_scalar.fit_transform(X), columns = breast_cancer_data.feature_names)
scaled_X
```

```python
X.mean(), X.std()
```

```python
scaled_X.mean(),scaled_X.std()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(scaled_X,y,test_size = 0.20,shuffle=True,random_state = 12,stratify=y)
```

```python
print("Training samples:", X_train.shape, y_train.shape)
print("Testing samples:", X_test.shape, y_test.shape)
```

```python
%%time
from sklearn.tree import DecisionTreeClassifier
Scaled_DT_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight = {0:2,1:1})
Scaled_DT_model
```

```python
Scaled_DT_model.fit(X_train,y_train)
```

```python
y_pred_scaled = Scaled_DT_model.predict(X_train)
```

```python
confusion_matrix(y_train,y_pred_scaled)
```

```python
print(classification_report(y_train,y_pred_scaled))
```

```python
y_pred_scaled_test = Scaled_DT_model.predict(X_test)
```

```python
confusion_matrix(y_test,y_pred_scaled_test)
```

```python
print(classification_report(y_test,y_pred_scaled_test))
```
