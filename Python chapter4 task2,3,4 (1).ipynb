##  merge() based on columns

```python
import pandas as pd
```

```python
#two dataframes are created

term1_marks = pd.DataFrame({'names':["Riya","Priya","Raju","Rambo"],
                 'math':[34,34,21,54],
                 'science':[32,54,65,76],
                 'ela':[43,65,76,87],
                 'comp':[33,43,76,87]})
term2_marks = pd.DataFrame({'names':["Riya","Priya","Raju","Rambo"],
                 'math':[54,76,87,98],
                 'science':[54,76,97,59],
                 'ela':[54,85,96,97],
                 'comp':[93,63,96,77]})

#merge function is carried out based on names column and inner merge operation is stated, which is also a default value
# inner merge returns the result dataframe only for the entries present in both the dataframes

result = term1_marks.merge(term2_marks, on = 'names', how = 'inner')
```

```python
#both the dataframes contain same values on column specified for merge, so all the values are added as extra columns in the result dataframe
result
```

```python
result11 = pd.merge(term1_marks,term2_marks, on = 'names', how = 'outer')
result11
```

```python
#returns all the values in both the dataframes NaN returned for entries not present in other dataframe

result1 = term1_marks.merge(term2_marks, on = 'names', how = 'outer')
result1
```

```python
result22 = pd.merge(term1_marks,term2_marks, on = 'names', how = 'left')
result22
```

```python
result2 = term1_marks.merge(term2_marks, on = 'names', how = 'left')
result2
```

```python
result3 = term1_marks.merge(term2_marks, on = 'names', how = 'right')
result3
```

### List comprehension - nested loops of series of conditions in a single line

```python
### sum vales calculated for range of numbers corresponding to the length

Internal_marks = [20,19,20,18]
Theory_marks = [76,78,76,79]

Total_marks =[[Internal_marks[i] + Theory_marks[i] for i in range(len(Internal_marks))]]

Total_marks
```

#### same as above executed using map() in lambda

```python
## using map() in lambda

Internal_marks = [20,19,20,18]
Theory_marks = [76,78,76,79]

Total_marks = list(map(lambda x, y: x + y, Internal_marks, Theory_marks))

Total_marks
```

```python
## create a list of grade_2 level salary from the given mixed list of salary

salary = [18000, 20000, 30000, 34000, 50000]
grade_2_salary = [a for a in salary if a >= 25000]
grade_2_salary
```

#### same as above executed using filter in lambda 

```python
salary = [48000, 20000, 30000, 34000, 50000,21000]
grade_2_salary = list(filter(lambda a: a >= 25000, salary))
grade_2_salary
```

### reduce in lambda 

```python
from functools import reduce

no_of_orders = [21,43,76,98,87,54]
total_orders = reduce(lambda x, y: x + y, no_of_orders)
total_orders
```
