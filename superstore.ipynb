```python
## 28.	Are there any states where sales are high but profit is low? 
```

```python
sales_profit_bystate = superstore_dataset.groupby('State/Province')[['Sales','Profit']].sum().sort_values('Sales', ascending=False)
sales_profit_bystate
```

```python
plt.title('high sales yet low profit generating states')
plt.xlabel('Sales')
plt.ylabel('Profit')
color = ['red' if sales > 75000 and profit < 0 else 'blue'
    for sales,profit in zip(
        sales_profit_bystate['Sales'],
        sales_profit_bystate['Profit']
    )]
plt.scatter(x = sales_profit_bystate['Sales'], y =sales_profit_bystate['Profit'], c= color)
plt.show()
```

```python
sales_profit_by_state = superstore_dataset.groupby('State/Province')[['Sales','Profit']].sum().sort_values('Profit', ascending=True).head(4)
sales_profit_by_state
```

```python
plt.title('high sales yet low profit generating states')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.scatter(x = sales_profit_by_state['Sales'], y =sales_profit_by_state['Profit'])
plt.show()
```

```python
## 29.	Are there any products where sales are high but the company is losing money? 
```

```python
low_profit = superstore_dataset.groupby('Product Name')[['Sales','Profit']].sum().sort_values('Sales', ascending=False).astype('int')
low_profit
```

```python
profit_byproduct = superstore_dataset.groupby('Product Name')[['Sales','Profit']].sum().sort_values('Profit', ascending=True).astype('int').head(12)
profit_byproduct
```

```python
plt.title('Loss generating High sales product')
plt.xlabel('Sales')
plt.ylabel('Profit')
colors = ['red' if profit < -2000 else 'blue'
         for profit in low_profit['Profit']
         ]
plt.scatter(x = low_profit['Sales'], y= low_profit['Profit'], c= colors)
plt.show()
```
