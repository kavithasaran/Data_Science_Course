# CCTV \ WEBCAM detection

```python
* Step 1: Import libraries
* Step 2: Import Model
* Step 3: Import Image/Video
* Step 4: Convert this into GreyScale
* Step 5: Use MultiscaleDetector to get the position co-ordinates
* Step 6: Use Looping statement to draw the Rectangular bounding box as much objects we get
* Step 7: Display the Image
* Step 8: Drop the Waitkey to Pause the Program for a given second, refresh it again and again, and get the Key to close
* Step 9: Kill the video/image frame
* Step 10: Close all the windows
```

```python
import cv2
```

```python
front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")

video = cv2.VideoCapture(0)


while True:
    success, frame = video.read()
    if success == True:
        grey_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(grey_image)
        print(face_detector)
    
        for x,y,w,h in face_detector:
            cv2.rectangle(grey_image,(x,y),(x+w,y+h),(255,0,0))
            
        cv2.imshow("image_window", grey_image)
        key = cv2.waitKey(1)
        if key == 81 or key == 113:
            break
        
video.release()
cv2.destroyAllWindows()
```
