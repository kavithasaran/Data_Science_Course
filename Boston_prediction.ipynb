# 1. Import necessary libraries

```python
import keras
import numpy as np
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.datasets import boston_housing
from keras.models import Sequential
from keras.layers import Dense
```

# 2. Load Dataset

```python
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()
```

# 3.Data understanding

```python
X_train.shape
```

```python
print("sample to test later manually")
print(X_test[3])
print(y_test[3])
print("===========================================")
print("data understanding")
print(X_train.shape, X_test.shape)
print(y_train.shape,y_test.shape)
print(y_train.min(), y_train.max())
print(y_test.min(), y_test.max())
```

# 4. Data preparation

```python
print("mean and std of train data:")
X_train.mean().round(1), X_train.std().round(1)
```

```python
print("mean and std of test data:")
X_test.mean().round(1), X_test.std().round(1)
```

```python
print("max and min value of test data:")
y_train.max(), y_test.min()
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print("mean and std of train data:")
X_train.mean().round(1), X_train.std().round(1)
```

```python
print("mean and std of test data:")
X_test.mean().round(1), X_test.std().round(1)
```

# 5. Model building

```python
model = Sequential()
model.add(layer = Dense(units = 64,activation='relu',input_shape=(13,)))
model.add(layer= Dense(units=32, activation='relu'))
model.add(layer= Dense(units=16, activation='relu'))

model.add(layer= Dense(units=1,))
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
model.summary()
```

# 6. model training | testing | evaluation

```python
model.fit(x=X_train,y=y_train,batch_size=16,epochs=100,verbose='auto',validation_split=0.20)
```

```python
y_pred = model.predict(X_test,batch_size=16,verbose='auto')
```

```python
y_test[3]
```

```python
y_pred[3]
```

```python
test_loss, test_mae = model.evaluate(x=X_test,y=y_test)
print("Testing Mean Absolute Error:", round((test_mae),2))
```
