## 4.1 correlation finding

```python
movie_pivot = pd.pivot_table(data = movie_data,values="rating",index="movie",columns="userId",fill_value=0)
movie_pivot.columns = movie_data["userId"].unique()
movie_pivot
```

```python
movie_pivot.corr().round(3)
```

## 4.2 building ubcf using Euclideon distance

```python
from sklearn.metrics import pairwise_distances
```

```python
pairwise_pivot =pd.pivot_table(data = movie_data,values="rating",index="userId",columns="movie",fill_value=0)
pairwise_pivot.index = movie_data["userId"].unique()
pairwise_pivot
```

```python
y = pd.DataFrame(pairwise_distances(X = pairwise_pivot,metric='euclidean'))
y.columns =  movie_data["userId"].unique()
y.index =  movie_data["userId"].unique()
y
```

## 4.3 Building UBCF using cosine similarity

```python
y_cosine = pd.DataFrame(1- pairwise_distances(X = pairwise_pivot,metric='cosine'))
y_cosine.columns =  movie_data["userId"].unique()
y_cosine.index =  movie_data["userId"].unique()
y_cosine
```
