```python
# Merge Category and Sub Category
Promotion_Table = Promotion_Table.merge(
    Category_Mapping_Table[['Product Line', 'SKU', 'Category', 'Sub Category']],
    on=['Product Line', 'SKU'],
    how='left'
)

# Merge MAP
Promotion_Table = Promotion_Table.merge(
    Price_List_Table[['Product Line', 'SKU', 'MAP']],
    on=['Product Line', 'SKU'],
    how='left'
)

# Display the updated Promotion_Table
Promotion_Table.head()
```

```python
Promotion_Table.head()
```

```python
Promotion_Table["SKU_Eligible"].value_counts()
```

```python
def calculate_promotion(row):

    # 1. SKU must be eligible
    if not row["SKU_Eligible"]:
        return 0

    # 2. Find matching offer rules
    matching_rules = Offer_Rules[
        (Offer_Rules["Season"] == row["Season"]) &
        (Offer_Rules["Category"] == row["Category"])
    ]

    # 3. Check sub-category
    matching_rules = matching_rules[
        (matching_rules["Sub Category"].isna()) |
        (matching_rules["Sub Category"] == row["Sub Category"])
    ]

    # 4. No matching promotion
    if matching_rules.empty:
        return 0

    # 5. Get the matching offer
    offer = matching_rules.iloc[0]

    # 6. Calculate promotion
    if offer["Offer Type"] == "Percentage":
        return row["MAP"] * offer["Offer Value"]

    elif offer["Offer Type"] == "Fixed":
        return offer["Offer Value"]

    return 0
```

```python
Promotion_Table["Promotion"] = (
    Promotion_Table.apply(
        calculate_promotion,
        axis=1
    )
)
```

```python
import pandas as pd
from playwright.async_api import async_playwright
import re
import os
from datetime import datetime


# --------------------------------------------------
# 1. Amazon URLs
# --------------------------------------------------

urls = [
    "https://www.amazon.com/Laptop-Computer-N4020-128GB-Windows/dp/B0H5BTZFC1",
    "https://www.amazon.com/Lenovo-Business-Laptop-Processor-Storage/dp/B0HFJ9BNPL",
    "https://www.amazon.com/HP-LaserJet-M140w-Wireless-Printer/dp/B09MSPNYQ2",
    "https://www.amazon.com/Brother-HL-L2405W-Monochrome-Subscription-Replenishment/dp/B0CPL7HRQN",
    "https://www.amazon.com/HP-Toner-Cartridge-Black-W2110A/dp/B0843HSL56",
    "https://www.amazon.com/Brother-Genuine-Cartridge-TN830XL-Yield/dp/B0BX2G5K2G"
]


# --------------------------------------------------
# 2. Create screenshot folder
# --------------------------------------------------

os.makedirs("amazon_screenshots", exist_ok=True)


# --------------------------------------------------
# 3. Function to extract ASIN
# --------------------------------------------------

def extract_asin(url):

    match = re.search(r"/dp/([A-Z0-9]{10})", url)

    if match:
        return match.group(1)

    return None


# --------------------------------------------------
# 4. Async scraping function
# --------------------------------------------------

async def scrape_amazon_products(urls):

    records = []

    async with async_playwright() as p:

        browser = await p.chromium.launch(headless=True)

        page = await browser.new_page()

        # Loop through all 6 URLs
        for i, url in enumerate(urls, start=1):

            print(f"\nScraping product {i} of {len(urls)}")
            print("URL:", url)

            try:

                # ------------------------------------------
                # Open Amazon product page
                # ------------------------------------------

                await page.goto(
                    url,
                    wait_until="domcontentloaded",
                    timeout=60000
                )

                # Give Amazon some time to load dynamic content
                await page.wait_for_timeout(3000)


                # ------------------------------------------
                # SKU / ASIN
                # ------------------------------------------

                sku = extract_asin(url)


                # ------------------------------------------
                # Seller name
                # ------------------------------------------

                seller_name = None

                seller_selectors = [
                    "#sellerProfileTriggerId",
                    "#merchant-info a",
                    "#merchant-info",
                    "#tabular-buybox .tabular-buybox-text",
                    "#buybox .a-size-small"
                ]

                for selector in seller_selectors:

                    element = await page.query_selector(selector)

                    if element:

                        text = await element.inner_text()

                        if text.strip():

                            seller_name = text.strip()

                            break


                # ------------------------------------------
                # Advertised price
                # ------------------------------------------

                advertised_price = None

                price_selectors = [
                    "#corePrice_feature_div .a-price .a-offscreen",
                    "#corePriceDisplay_desktop_feature_div .a-price .a-offscreen",
                    "#priceblock_ourprice",
                    "#priceblock_dealprice",
                    ".a-price .a-offscreen"
                ]

                for selector in price_selectors:

                    element = await page.query_selector(selector)

                    if element:

                        text = await element.inner_text()

                        if text.strip():

                            advertised_price = text.strip()

                            break


                # ------------------------------------------
                # Region
                # ------------------------------------------

                region = "US"


                # ------------------------------------------
                # Marketplace
                # ------------------------------------------

                marketplace = "Amazon.com"


                # ------------------------------------------
                # Last time collected
                # ------------------------------------------

                last_time_collected = datetime.now().strftime(
                    "%Y-%m-%d %H:%M:%S"
                )


                # ------------------------------------------
                # Screenshot
                # ------------------------------------------

                screenshot_path = (
                    f"amazon_screenshots/"
                    f"product_{i}_{sku}.png"
                )

                await page.screenshot(
                    path=screenshot_path,
                    full_page=True
                )


                # ------------------------------------------
                # Store result
                # ------------------------------------------

                records.append({

                    "Seller_name": seller_name,

                    "SKU": sku,

                    "Region": region,

                    "Advertised_price": advertised_price,

                    "Marketplace": marketplace,

                    "Screenshot": screenshot_path,

                    "Live_link": page.url,

                    "Last_time_collected": last_time_collected
                })


                print("Seller:", seller_name)
                print("SKU:", sku)
                print("Price:", advertised_price)
                print("Status: Success")


            except Exception as e:

                print("Error:", e)

                # Still create a row for this product
                records.append({

                    "Seller_name": None,

                    "SKU": extract_asin(url),

                    "Region": "US",

                    "Advertised_price": None,

                    "Marketplace": "Amazon.com",

                    "Screenshot": None,

                    "Live_link": url,

                    "Last_time_collected":
                        datetime.now().strftime(
                            "%Y-%m-%d %H:%M:%S"
                        )
                })


        # ------------------------------------------
        # Close browser
        # ------------------------------------------

        await browser.close()


    # ------------------------------------------
    # Create DataFrame
    # ------------------------------------------

    Seller_Table = pd.DataFrame(records)

    return Seller_Table
```

### seller files consolidation and cleaning

```python
import pandas as pd
from pathlib import Path
import re


# --------------------------------------------------
# 1. FOLDER LOCATION
# --------------------------------------------------

folder_path = Path(r"C:\Data Science/MAP Project/Seller Files")


# --------------------------------------------------
# 2. FUNCTION TO EXTRACT DATE FROM FILE NAME
# --------------------------------------------------

def extract_date_from_filename(file_name):

    # Looks for dates such as:
    # 2026-09-27
    # 2026_09_27
    # 2026.09.27

    match = re.search(
        r"(\d{4})[-_.](\d{2})[-_.](\d{2})",
        file_name
    )

    if match:
        date_string = "-".join(match.groups())
        return pd.to_datetime(date_string)

    return pd.NaT


# --------------------------------------------------
# 3. FUNCTION TO CLEAN EACH CSV FILE
# --------------------------------------------------

def clean_seller_file(file_path):

    # Read CSV
    df = pd.read_csv(file_path)

    # Clean column names
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    # Rename columns
    df = df.rename(columns={
        "Seller_name": "Seller_Name",
        "adv_price": "Advertised_Price",
        "Marketplace": "Marketplace",
        "region": "Region",
        "SKU": "SKU"
    })

    # Drop unwanted columns
    df = df.drop(
        columns=["Screenshot", "live_link"],
        errors="ignore"
    )

    # Remove completely empty rows
    df = df.dropna(how="all")

    # --------------------------------------------------
    # Clean text columns
    # --------------------------------------------------

    text_columns = [
        "SKU",
        "Marketplace",
        "Seller_Name",
        "Region"
    ]

    for col in text_columns:
        if col in df.columns:
            df[col] = (
                df[col]
                .astype("string")
                .str.strip()
            )

    # --------------------------------------------------
    # Clean Advertised Price
    # --------------------------------------------------

    if "Advertised_Price" in df.columns:

        df["Advertised_Price"] = (
            df["Advertised_Price"]
            .astype("string")
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
            .str.strip()
        )

        df["Advertised_Price"] = pd.to_numeric(
            df["Advertised_Price"],
            errors="coerce"
        )

    # --------------------------------------------------
    # Add Violation Date from filename
    # --------------------------------------------------

    df["Violation_Date"] = extract_date_from_filename(
        file_path.name
    )

    return df


# --------------------------------------------------
# 4. READ ALL CSV FILES
# --------------------------------------------------

csv_files = list(folder_path.glob("*.csv"))

seller_data = []

for file in csv_files:

    # IMPORTANT:
    # Don't process the files we create later
    if file.stem.startswith("seller_details_"):
        continue

    cleaned_df = clean_seller_file(file)

    seller_data.append(cleaned_df)


# --------------------------------------------------
# 5. COMBINE ALL FILES
# --------------------------------------------------

Seller_Table = pd.concat(
    seller_data,
    ignore_index=True
)


# --------------------------------------------------
# 6. FINAL COLUMN ORDER
# --------------------------------------------------

Seller_Table = Seller_Table[
    [
        "sku",
        "marketplace",
        "seller_name",
        "Region",
        "Advertised_Price",
        "Violation_Date"
    ]
]


# --------------------------------------------------
# 7. SPLIT BY REGION
# --------------------------------------------------

regions = ["USA", "UAE", "UK", "IND", "CAN"]

for region in regions:

    region_df = Seller_Table[
        Seller_Table["Region"].str.upper() == region
    ].copy()

    output_file = folder_path / f"seller_details_{region}.csv"

    region_df.to_csv(
        output_file,
        index=False
    )


# --------------------------------------------------
# 8. DISPLAY RESULTS
# --------------------------------------------------

print("Processing completed successfully.")

print("\nTotal records:", len(Seller_Table))

print("\nRecords by region:")
print(Seller_Table["Region"].value_counts())

print("\nFiles saved in:")
print(folder_path)
```

# 3. Violation table logic

## Create VIOLATION Table

```python
import pandas as pd
import numpy as np

# Select the required columns for the Violation Table
columns = [
    "SKU",
    "Marketplace",
    "Advertised Seller Name",
    "Region",
    "Violation_Date",
    "Sub Category",
    "Actual Seller"
]

Violation_Table = Price_Monitoring_Table[columns].copy()

# Add pricing fields temporarily for calculating violations
Violation_Table["Advertised Price"] = pd.to_numeric(
    Price_Monitoring_Table["Advertised_Price"], errors="coerce"
)
Violation_Table["MAP"] = pd.to_numeric(
    Price_Monitoring_Table["MAP"], errors="coerce"
)
Violation_Table["LPP_Price"] = pd.to_numeric(
    Price_Monitoring_Table["LPP_Price"], errors="coerce"
)
Violation_Table["Promotion"] = pd.to_numeric(
    Price_Monitoring_Table["Promotion"], errors="coerce"
)

# Calculate the promotional allowed limit
Violation_Table["Allowed_Limit"] = (
    Violation_Table["MAP"] - Violation_Table["Promotion"]
)

# Initialize the three policy columns
Violation_Table["Policy Compliant"] = "NO"
Violation_Table["Promotion Violation"] = "NO"
Violation_Table["LPP Violation"] = "NO"

# Validate required pricing data before classification
required = ["Advertised Price", "MAP", "LPP_Price", "Promotion"]

if Violation_Table[required].isna().any().any():
    raise ValueError(
        "Missing or invalid price data found. "
        "Please clean the affected rows before continuing."
    )

# Rule 1: Advertised price is at or above MAP
mask = Violation_Table["Advertised Price"] >= Violation_Table["MAP"]
Violation_Table.loc[mask, "Policy Compliant"] = "YES"

# Rule 2: Promotion exists and price is within the allowed limit
mask = (
    (Violation_Table["Promotion"] > 0) &
    (Violation_Table["Advertised Price"] < Violation_Table["MAP"]) &
    (Violation_Table["Advertised Price"] >= Violation_Table["Allowed_Limit"])
)
Violation_Table.loc[mask, "Policy Compliant"] = "YES"

# Rule 3: Promotion exists and price is below allowed limit,
# but is not below LPP
mask = (
    (Violation_Table["Promotion"] > 0) &
    (Violation_Table["Advertised Price"] < Violation_Table["Allowed_Limit"]) &
    (Violation_Table["Advertised Price"] >= Violation_Table["LPP_Price"])
)
Violation_Table.loc[mask, "Promotion Violation"] = "YES"

# Rule 4: Promotion exists and price is below both limits
mask = (
    (Violation_Table["Promotion"] > 0) &
    (Violation_Table["Advertised Price"] < Violation_Table["Allowed_Limit"]) &
    (Violation_Table["Advertised Price"] < Violation_Table["LPP_Price"])
)
Violation_Table.loc[mask, "Promotion Violation"] = "YES"
Violation_Table.loc[mask, "LPP Violation"] = "YES"

# Rule 5: No promotion, but price is below MAP and at/above LPP
mask = (
    (Violation_Table["Promotion"] == 0) &
    (Violation_Table["Advertised Price"] < Violation_Table["MAP"]) &
    (Violation_Table["Advertised Price"] >= Violation_Table["LPP_Price"])
)
Violation_Table.loc[mask, "Policy Compliant"] = "YES"

# Rule 6: No promotion and price is below LPP
mask = (
    (Violation_Table["Promotion"] == 0) &
    (Violation_Table["Advertised Price"] < Violation_Table["LPP_Price"])
)
Violation_Table.loc[mask, "LPP Violation"] = "YES"

# Check the result
display(Violation_Table.head(10))
```

## ADD counter to track the repeated violation

```python
# Convert violation dates to datetime
Violation_Table["Violation_Date"] = pd.to_datetime(
    Violation_Table["Violation_Date"],
    errors="coerce"
)

# Sort chronologically
Violation_Table = Violation_Table.sort_values(
    ["Violation_Date"],
    kind="stable",
    na_position="last"
).reset_index(drop=True)

# Count promotion violations independently
Violation_Table["Promotion Violation Count"] = (
    Violation_Table.groupby(
        ["Actual Seller", "SKU"]
    )["Promotion Violation"]
    .transform(lambda s: s.eq("YES").cumsum())
)

# Count LPP violations independently
Violation_Table["LPP Violation Count"] = (
    Violation_Table.groupby(
        ["Actual Seller", "SKU"]
    )["LPP Violation"]
    .transform(lambda s: s.eq("YES").cumsum())
)

# Determine the action for each observation
Violation_Table["Recommended Action"] = "No Action"

violation_mask = (
    (Violation_Table["Promotion Violation"] == "YES") |
    (Violation_Table["LPP Violation"] == "YES")
)

suspension_mask = (
    (Violation_Table["Promotion Violation Count"] >= 3) |
    (Violation_Table["LPP Violation Count"] >= 3)
)

Violation_Table.loc[violation_mask, "Recommended Action"] = (
    "Warning Letter"
)

Violation_Table.loc[suspension_mask & violation_mask,
                    "Recommended Action"] = "SKU Suspension Letter"

display(Violation_Table.head(5))
```

```python
import pandas as pd
from docx import Document
from docx.shared import Inches, Pt
from pathlib import Path
import re

# Manufacturer name
MANUFACTURER = "Manufacturing Company"

# Output folder
output_folder = Path(
    "C:/Data Science/MAP Project/MAP Violation letters"
)
output_folder.mkdir(parents=True, exist_ok=True)

letters_created = 0
warning_count = 0
suspension_count = 0

# Helper function to handle missing values
def clean_value(value):
    if pd.isna(value):
        return "Not available"
    return str(value).strip()


# Select observations with a pricing violation
letters = Violation_Table[
    (
        pd.to_numeric(
            Violation_Table["LPP Violation Count"],
            errors="coerce"
        ).fillna(0) > 0
    )
    |
    (
        pd.to_numeric(
            Violation_Table["Promotion Violation Count"],
            errors="coerce"
        ).fillna(0) > 0
    )
].copy()


# Generate one separate Word document per observation
for observation_number, (_, row) in enumerate(
    letters.iterrows(), start=1
):

    # Read violation counts
    lpp_count = pd.to_numeric(
        pd.Series([row["LPP Violation Count"]]),
        errors="coerce"
    ).fillna(0).iloc[0]

    promotion_count = pd.to_numeric(
        pd.Series([row["Promotion Violation Count"]]),
        errors="coerce"
    ).fillna(0).iloc[0]

    # Determine violation type
    violation_types = []

    if lpp_count > 0:
        violation_types.append("MAP Violation")

    if promotion_count > 0:
        violation_types.append("Promotion Violation")

    violation_type = " and ".join(violation_types)

    # More than 3 violations of either type triggers suspension
    if lpp_count > 3 or promotion_count > 3:
        action = "Suspension"
        title = "SKU SUSPENSION NOTICE"
        subject = (
            f"SKU Suspension Notice - {violation_type}"
        )
        suspension_count += 1

    else:
        action = "Warning"
        title = "WARNING NOTICE - PRICING POLICY VIOLATION"
        subject = (
            f"Warning Notice - {violation_type}"
        )
        warning_count += 1

    # Create a NEW Word document for this observation
    doc = Document()

    # Page margins and default font
    section = doc.sections[0]
    section.top_margin = Inches(0.7)
    section.bottom_margin = Inches(0.7)

    doc.styles["Normal"].font.name = "Arial"
    doc.styles["Normal"].font.size = Pt(10)

    # Basic information
    sku = clean_value(row["SKU"])
    actual_seller = clean_value(row["Actual Seller"])

    violation_date = row["Violation_Date"]

    if pd.notna(violation_date):
        date_text = pd.to_datetime(
            violation_date
        ).strftime("%B %d, %Y")
    else:
        date_text = "Date unavailable"

    # Letter heading
    doc.add_heading(MANUFACTURER, level=1)
    doc.add_paragraph(f"From: {MANUFACTURER}")
    doc.add_paragraph(f"To: {actual_seller}")
    doc.add_paragraph(f"Date: {date_text}")

    doc.add_heading(title, level=2)
    doc.add_paragraph(f"Subject: {subject}")

    # Violation details
    doc.add_heading("Violation Details", level=3)

    details = [
        ("SKU", sku),
        ("Violation Type", violation_type),
        ("Recommended Action", action),
        ("Marketplace", clean_value(row["Marketplace"])),
        (
            "Advertised Seller Name",
            clean_value(row["Advertised Seller Name"])
        ),
        ("Actual Seller", actual_seller),
        ("Region", clean_value(row["Region"])),
        ("Violation Date", date_text),
        ("Sub Category", clean_value(row["Sub Category"])),
        ("LPP Violation Count", int(lpp_count)),
        ("Promotion Violation Count", int(promotion_count)),
    ]

    table = doc.add_table(rows=0, cols=2)
    table.style = "Table Grid"

    for label, value in details:
        cells = table.add_row().cells
        cells[0].text = label
        cells[1].text = str(value)

    doc.add_paragraph()
    doc.add_paragraph("Dear Reseller,")

    # Letter content based on action
    if action == "Warning":

        if lpp_count > 0 and promotion_count > 0:
            message = (
                f"Our records indicate that SKU {sku} has "
                "pricing violations relating to both the "
                "applicable MAP policy and "
                "promotional pricing requirements. Please "
                "review the applicable pricing requirements "
                "and correct the advertised price as necessary."
            )

        elif lpp_count > 0:
            message = (
                f"Our records indicate that the advertised "
                f"price for SKU {sku} is below the applicable "
                "Lowest Permitted Price (LPP), constituting "
                "a MAP policy violation under the applicable "
                "pricing rules. Please correct the advertised "
                "price and ensure future compliance."
            )

        else:
            message = (
                f"Our records indicate that the advertised "
                f"price for SKU {sku} does not comply with "
                "the applicable promotional pricing "
                "requirements. Please review the applicable "
                "promotion and correct the advertised price "
                "as necessary."
            )

        doc.add_paragraph(message)

        doc.add_paragraph(
            "This notice serves as a warning. Please take "
            "the required corrective action and ensure "
            "future compliance with the applicable "
            "manufacturer pricing policies."
        )

    else:
        # Suspension notice
        reasons = []

        if lpp_count > 3:
            reasons.append(
                f"more than three LPP-related violations "
                f"({int(lpp_count)} recorded)"
            )

        if promotion_count > 3:
            reasons.append(
                f"more than three promotion violations "
                f"({int(promotion_count)} recorded)"
            )

        reason_text = " and ".join(reasons)

        doc.add_paragraph(
            f"Repeated pricing violations have been recorded "
            f"for SKU {sku}. The recorded violation history "
            f"meets the suspension threshold: {reason_text}."
        )

        doc.add_paragraph(
            f"You are hereby notified that selling rights for "
            f"SKU {sku} are subject to suspension in accordance "
            "with the applicable manufacturer agreement and "
            "enforcement process. Please discontinue selling "
            "this specific SKU as directed by the manufacturer "
            "under the applicable agreement."
        )

        doc.add_paragraph(
            "This notice applies only to the identified SKU. "
            "Other products from the manufacturer may continue "
            "to be sold in accordance with the applicable "
            "agreement and pricing policies."
        )

    doc.add_paragraph(
        "Please review the applicable manufacturer pricing "
        "policy and take the required corrective action."
    )

    doc.add_paragraph("Sincerely,")
    doc.add_paragraph(
        f"Compliance Team\n{MANUFACTURER}"
    )

    # Create a unique filename for every observation
    safe_seller = re.sub(
        r'[<>:"/\\\\|?*]', "_", actual_seller
    )
    safe_sku = re.sub(
        r'[<>:"/\\\\|?*]', "_", sku
    )

    output_file = output_folder / (
        f"Notice_{observation_number:04d}_"
        f"{safe_seller}_{safe_sku}.docx"
    )

    # Save this individual document immediately
    doc.save(output_file)
    letters_created += 1

    print(f"Saved: {output_file.name}")


# Final summary
print("\nLetter generation completed.")
print(f"Total individual letters created: {letters_created}")
print(f"Warning letters: {warning_count}")
print(f"Suspension letters: {suspension_count}")
print(f"Output folder: {output_folder.resolve()}")

```
