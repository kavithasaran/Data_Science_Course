# CCTV \ WEBCAM detection

```python
* Step 1: Import libraries
* Step 2: Import Model
* Step 3: Import Image/Video
* Step 4: Convert this into GreyScale
* Step 5: Use MultiscaleDetector to get the position co-ordinates
* Step 6: Use Looping statement to draw the Rectangular bounding box as much objects we get
* Step 7: Display the Image
* Step 8: Drop the Waitkey to Pause the Program for a given second, refresh it again and again, and get the Key to close
* Step 9: Kill the video/image frame
* Step 10: Close all the windows
```

```python
import cv2
from ultralytics import YOLO
import pandas as pd

all_detections = []

model = YOLO("yolov8n.pt")

video = cv2.VideoCapture(0)


while True:
    success, frame = video.read()
    if success == True:
        results = model(frame)
        print(results)

        for result in results:
            boxes = result.boxes
            for box in boxes:
                x1,y1,x2,y2 = map(int, box.xyxy[0])
                confidence = float(box.conf[0])
                class_id = int(box.cls[0])
                class_name = model.names[class_id]
                all_detections.append({"Object": class_name, "Confidence": confidence})
                cv2.rectangle(frame,(x1,y1),(x2,y2),(0,255,0),3)
                label = f"{class_name} {confidence:.2f}"
                cv2.putText(frame,label,(x1,y1-10), cv2.FONT_HERSHEY_SIMPLEX,0.7,(0,255,0),2)
                
        df_results = pd.DataFrame(all_detections)
        df_results["Confidence"] = (df_results["Confidence"] * 100).round(2)
        cv2.imshow("YOLO Webcam Detection", frame)
        key = cv2.waitKey(1)
        if key == 81 or key == 113:
            break
    else:
        print("Video Stopped")
print(df_results)        
video.release()
cv2.destroyAllWindows()
```
