## data preparation, model building, model training

### phase 1-------- core 4 step setup stage 

####  4.1 Initialize weights randomly

```python
w = np.random.randn(1,1)
b = np.random.randn(1,1)

print("Initial weight : ", w)
print("Initial Bias : ", b)
```

#### 4.2 Forward pass prediction

```python
def forward(X, w, b):
    return np.dot(X, w) + b
y_pred = forward(X, w, b)
print("Predictions: \n", y_pred)
```

#### 4.3 loss function

```python
def loss(y, y_pred):
    return np.mean((y-y_pred) ** 2)
print("Loss:", loss(y,y_pred))
```

#### 4.4 backward propagation (manual gradient)

```python
def gradients(X, y, y_pred):
    n = len(X)

    dw = (-2/n) * np.dot(X.T, (y-y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    return dw, db
```

### phase 2 - the actual training phase

```python
print("Before Training:", forward(np.array([[5]]), w, b))
```

```python
learning_rate = 0.01

for i in range(1000):
    y_pred = forward(X, w, b)

    dw, db = gradients(X, y, y_pred)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    if i % 100 == 0:
        print(f"Epoch {i}, loss: {loss(y, y_pred)}")
```

```python
print("After training:", forward(np.array([[5]]), w, b))
```

### so learning = updating weight and bias to reduce error
